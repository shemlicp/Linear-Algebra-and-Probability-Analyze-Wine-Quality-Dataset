```python
import pandas as pd
import numpy as np

# 1. Load the dataset and handle missing data
# Change sep=';' to sep=',' if your specific file uses standard commas
df = pd.read_csv('winequality-red.csv', sep=';') 

# Replace any null values with the mean of their respective column
df.fillna(df.mean(numeric_only=True), inplace=True)
print("1. Missing values after cleaning:\n", df.isnull().sum(), "\n")

# 2. Extract 'alcohol' and 'citric acid' columns as vectors (1D NumPy arrays)
alcohol_vector = df['alcohol'].to_numpy()
citric_acid_vector = df['citric acid'].to_numpy()

print("2. Vector Shapes:")
print("Alcohol vector shape:", alcohol_vector.shape)
print("Citric acid vector shape:", citric_acid_vector.shape, "\n")

# 3. Select two features and calculate the covariance matrix
X = df[['alcohol', 'density']].to_numpy()
cov_matrix = np.cov(X.T)

print("3. Covariance Matrix (alcohol vs. density):\n", cov_matrix, "\n")

# 4. Perform Eigen Decomposition on the covariance matrix
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Sort eigenvalues and eigenvectors in descending order (top to lowest)
sorted_indices = np.argsort(eigenvalues)[::-1]
top_eigenvalues = eigenvalues[sorted_indices]
top_eigenvectors = eigenvectors[:, sorted_indices]

print("4. Eigen Decomposition:")
print("Top 2 Eigenvalues:\n", top_eigenvalues)
print("Corresponding Eigenvectors (as columns):\n", top_eigenvectors, "\n")

# 5. Find the most common wine quality and its distribution
most_common_quality = df['quality'].mode()[0]
quality_counts = df['quality'].value_counts().sort_index()
quality_probabilities = df['quality'].value_counts(normalize=True).sort_index() * 100

print(f"5. Most Common Wine Quality: {most_common_quality}")
print("\nQuality Score Counts:\n", quality_counts)
print("\nQuality Score Percentages (%):\n", quality_probabilities.round(2))
```
